```python
# PRACTICAL 8
# Mini AI Application: Text Classification

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

# Training data
messages = [
    "Congratulations you won a free prize",
    "Claim your free gift now",
    "Win cash by clicking this link",
    "Limited offer buy now",
    "You have won a lottery",
    "Hello how are you",
    "Please attend the class tomorrow",
    "Your assignment is due on Monday",
    "Can we meet for lunch today",
    "The meeting starts at ten"
]

# Labels
labels = [
    "Spam",
    "Spam",
    "Spam",
    "Spam",
    "Spam",
    "Not Spam",
    "Not Spam",
    "Not Spam",
    "Not Spam",
    "Not Spam"
]

# Create the AI model
model = make_pipeline(
    CountVectorizer(),
    MultinomialNB()
)

# Train the model
model.fit(messages, labels)

# Test messages
test_messages = [
    "You won a free gift",
    "Please attend the meeting today",
    "Claim your cash prize now"
]

# Predict message categories
predictions = model.predict(test_messages)

# Display results
print("AI TEXT CLASSIFICATION - SPAM DETECTION")
print("---------------------------------------")

for message, prediction in zip(test_messages, predictions):
    print("Message:", message)
    print("Prediction:", prediction)
    print()
```
