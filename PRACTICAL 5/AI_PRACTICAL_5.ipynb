# Build a small rule-based expert system or logic inference engine fora diagnosis or advisory problem

```python
def rule_based(turn_on,internet,slow):
  if turn_on == "no":
    print("no charging in your laptop")
  elif internet == "no":
    print("reconnect your wifi")
  elif slow == "yes":
    print("restart your divice")
  else:
    print("your laptop is charged and internet is connected")

turn_on = input("does the laptop is on? YES/NO").lower()
internet = input("wifi is connected or not YES/NO").lower()
slow = input("is your laptop is slow or notYES/NO").lower()
advice = rule_based(turn_on,internet,slow)
print(advice)

```
