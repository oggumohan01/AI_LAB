```python
print("Hello! I am cleaning robo.")
print("Today we will build a cleaning robot together. 🤖")
print(2 + 3, "is the answer to 2 + 3")

# This is our room: 5 spots. D = dirty, C = clean
room = ["D", "C", "D", "D", "C"]

# A helper that draws the room with nice symbols
def show_room(room):
    picture = ""
    for spot in room:
        if spot == "D":
            picture += "💩 "
        else:
            picture += "✨ "
    print(picture)

print("Our room right now:")
show_room(room)

# The robot looks at ONE spot and decides what to do.
def clean_spot(spot):
    if spot == "D":
        return "C"
    else:
        return "C"

# Let's test the robot on one dirty spot:
result = clean_spot("D")
print("The robot looked at a dirty spot 💩 and made it:", result,
      "(C means clean ✨)")

print("BEFORE - the dirty room:")
show_room(room)

# The robot visits every spot, one by one
for i in range(len(room)):
    room[i] = clean_spot(room[i])
    print("After cleaning spot number", i + 1, ":")
    show_room(room)

print()
print("AFTER - all done! ⭐⭐⭐⭐⭐⭐")
```

### Visualizing the Cleaning Robot in Action

Let's visualize how our cleaning robot works by showing the room's state before and after each cleaning step.

```python
# Reset the room to its initial state for a fresh visualization
room = ["D", "C", "D", "D", "C"]

print("Initial room state:")
show_room(room)
print("--------------------")
```

The robot will now go through each spot, one by one. After cleaning each spot, we'll see the updated state of the room.

```python
# Simulate the robot cleaning each spot and visualize the progress
for i in range(len(room)):
    original_spot_state = room[i]
    room[i] = clean_spot(room[i])
    print(f"Robot cleans spot {i + 1}: Changed from {original_spot_state} to C")
    print(f"Room state after cleaning spot {i + 1}:")
    show_room(room)
    print("--------------------")

print("Final room state after all spots are cleaned:")
show_room(room)
print("All done! ⭐⭐⭐⭐⭐⭐")
```
